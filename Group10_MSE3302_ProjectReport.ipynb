# Miniature C-arm Simulator

## Abstract
C-arm Fluoroscopy machines are an essential part of guiding medical professionals to visualize procedures. Real time imaging ensures accuracy while observing internal systems for diagnosis, inspection and technique performance. However, their manufacturing size and cost as well as radiation risks make them impractical for training and selling purposes. Based on the accelerometer sensors returning position-based imaging of a human spine, our miniature model approach accurately depicts this technology, without the need for extensive transportation methods, radiation, setup or risk of damaging real equipment. 

The design includes 3 degrees of freedom using three stepper motors, an Arduino Nano as the core control and wireless communication to a screen with preset or exact positions. With the goals of optimizing portability and repeatability for training with limited resources, the structure is completely 3D-printable and designed for easy assembly. 

The project demonstrates that training or business pitching of expensive machinery does not need to require full scale integrations of systems. By eliminating the reliance from clinical material, opportunities for inclusivity arise.   


## Introduction

C-arms are medical imaging systems that use fluoroscopy to produce real-time X-ray vision. They utilize small amounts of ionizing radiation for diagnostic, therapeutic, or research purposes, minimizing radiation exposure to patients and workers. This approach aims to achieve function while reducing the risk of harmful radiation effects. Procedures like Spinal Cord Stimulator Implantation, for patients with chronic pain from nerve damage or unsuccessful back surgery, require millimeter-level precision to ensure full recovery. This procedure involves delivering low electrical pulses to specific areas of the spinal cord to block pain signals from reaching the brain. A lack of precision can lead to nerve root damage, causing paralysis, loss of sensation or motor function in limbs, meningitis, and death. Incorporating a C-arm device with real-time visualization helps avoid misplacement, verify final placements, and reduce revision surgeries.

The ALARA (As Low As Reasonably Achievable) principle dictates that radiation exposure should be as minimal as possible without compromising necessary outcomes. To reduce DNA and tissue damage, as well as other long-term effects, using the lowest dose settings and employing filters is necessary. When training is required, the additional and unnecessary exposure to radiation introduces safety risks for all personnel involved. Additionally, standard training procedures for utilizing a C-arm fluoroscopy device involve expensive equipment and are impractical for frequent training sessions.

To effectively portray C-arm use, transportation of this machine from different medical facilities would require moving a 700-1000lb machine, standing on average over 7ft high and 6.5ft long. Moving this large and delicate piece of equipment is not only physically unfeasible but also poses a risk to equipment calibration, sterility, and structural integrity. Additionally, many C-arm machines require specific electrical and radiation safety setups that are difficult to replicate outside of a standard clinical environment.

These constraints highlight the importance of mock units in showcasing the device's functionality. To mitigate the logistical challenges of providing insightful demonstrations of these devices, a reduced model of the technology was 3D-printed, wired, and automated. Custom software was developed to generate real-time Digitally Reconstructed Radiographs (DRRs) of the theoretical medical imaging that practitioners would visualize during regular device use. This approach successfully eliminates unnecessary radiation exposure while emphasizing the advantageous services offered by the mechanism. he scaled down model effectively weighs approximately 947g. While being compact, spanning from 200cm in height and 240 in length, it is designed for easy assembly and disassembly. This allows for efficient transportation and fast reassembly, making it ideal for educational and demonstrational purposes. Tracking is achieved via stepper motors for precise movement control, while integrated accelerometers link the physical system and the software, synchronizing the movement to the image displayed. This design combines cost-efficiency and precision.

#### 6 Degrees of Freedom
Incorporating as many degrees of freedom (DOF) as possible was identified as critical for creating a functional training application. Conversations with medical professionals emphasized the importance of replicating the full range of motion a real C-arm offers. However, due to limitations in scope and complexity, we opted to include as many DOF as feasible within the constraints of our project.

#### Proximity Sensors
Proximity sensors enhance realism but add unnecessary complexity and assembly difficulty, conflicting with our goals of simplicity and accessibility.

#### Foot Pedal Activation
Including a foot pedal would add bulk and complicate the assembly, making it impractical given our focus on streamlined manufacturing.

#### Mobile Application Control
A mobile app offers incredibly accessible and streamlined operation, aligning well with our goals of simplicity and ease of use.

#### Voice Activation
Voice control is overly complex and difficult to implement, making it impractical for this project.

#### Preset Starting Positions
Preset positions simplify training by allowing quick resets, enhancing consistency and usability.

Based on these findings, our design will implement three degrees of freedom, which provides sufficient functionality for training without overcomplicating the model. It will utilize mobile application control for streamlined operation and include preset starting positions to enhance training consistency and usability.


### Mechanical/Component Design Methods
The guiding principles for our mechanical design were simplicity and 3D-printability. Our goal was to ensure the entire model could be manufactured using accessible, affordable methods without compromising the intended functionality. We began by selecting components that offered the most practical balance of cost-effectiveness and ease of integration. From these selections, we developed a concept for the structural framework and mechanical systems of the miniature C-arm. Using SOLIDWORKS, we iteratively refined the design to achieve our desired final product.

![Drawing](https://github.com/KaiElrick/3302Proj/blob/main/Screenshot%202025-04-03%20124422.png?raw=true)

Building upon the initial concept, we designed and assembled the C-arm with three degrees of freedom to allow for a realistic range of motion. The primary components included:

#### Linear Motion Table & Base
Based on a simple Google search, "A typical C-arm table is approximately 24" wide and 84" long." Using this as a baseline, we adopted an approximate 3:1 length-to-width scale, with the final model being scaled approximately 11 times smaller than the original.

The table was designed as a simple flat plate with a dovetail slider to allow for smooth linear motion and precise positioning. The dovetail profile was chosen for its compact size, self-locking geometry, and simplicity. The table base, containing the dovetail saddle, was cantilevered out to accommodate the C-arm, providing sufficient clearance for rotation and movement. Two support beams were incorporated to securely hold the motor and enhance structural stability. Components were intentionally left exposed to simplify assembly, maintenance, and accessibility. The table's height was selected to align with the center of the C-arm's rotation, ensuring proper functionality and realistic operation.

A rack and pinion mechanism was used to enable translational movement of the table. The table base was designed so that the motor could be attached at any height during assembly, allowing for precise meshing of the gears without extensive testing to ensure alignment. Similarly, the rack was printed separately from the table to allow alignment with the pinion during assembly. This method was necessary due to the tolerances of the 3D printer (greater than 0.2mm), which made precise gear meshing unpredictable. The table base was too large to print iteratively.

#### Rotational Drive Gears
The rotational drive gears were designed to provide controlled rotational motion of the C-arm, accurately simulating the maneuverability of a real fluoroscopy machine. To streamline construction and enhance simplicity, we used a uniform module between all gears in our design, applicable to both rotational mechanisms. This approach reduced complexity, making assembly easier and ensuring consistency across the model.

Due to limitations in our 3D-printing capabilities, the gears were made larger than ideal. They were designed with a module of 2mm and a 20-degree pitch angle, optimized to be driven by a stepper motor and consistently provide accurate, repeatable outputs. The gears were integrated into the C-arm support structure, allowing for smooth and consistent rotation about a central axis.

#### C-Arm Assembly
The C-arm assembly was designed to closely mimic the structure and functionality of a real fluoroscopy machine. To replicate these capabilities at a reduced scale, we constructed our C-arm with an inner diameter of approximately 144 mm (6 inches). Given that a standard full-size C-arm has a diameter of about 65 inches, this achieves a scale ratio of roughly 11:1, consistent with the overall model. The C-arm features a continuous curved structure to provide unobstructed rotational motion around the patient area, effectively replicating the movement capabilities of a real C-arm.

The C-arm was modeled as a large gear for drive purposes and used the global module of 2mm. A dovetailer profile slider was used to support the C-arm. A base was created on one end of the C-arm to hold the accelerometer.

#### C-Arm Support Bracket
The C-Arm support bracket holds the C-Arm, as well as the gear and motor that drive its rotational degree of freedom. The bracket was designed slightly larger than scale to accommodate a stepper motor that cannot be scaled down. The bracket mounts directly to a stepper motor on the main stand, allowing the whole assembly to rotate.

#### Electrical Component Choices Justification
Our design approach prioritized cost-effective, reliable, and easily replicable solutions to create a practical training tool. By utilizing 3D printing for most mechanical components, we ensured that the model could be easily reproduced and potentially open-sourced for broader accessibility. For the electrical components, we focused on selecting well-documented, mass-produced, and widely supported hardware to simplify implementation and troubleshooting.

Stepper motors were chosen over DC motors due to their precise position control, high torque at low speeds, and minimal vibration. The 28BYJ-48 stepper motor and ULN2003 driver board were selected for their availability, affordability, and compatibility with various microcontrollers. The ULN2003 motor driver simplifies the interface between the stepper motor and the control system, ensuring a compact and efficient solution that meets the project's performance needs.

#### Stepper Motor: 28BYJ-48
The 28BYJ-48 stepper motor was selected for its high precision, low power consumption, and cost-effectiveness. This unipolar stepper motor has a gear reduction ratio of 64:1, which significantly increases torque while maintaining fine control over movement.

**Technical Specifications:**
* Voltage: 5V
* Current Draw: ~240mA per phase
* Step Angle: ~5.625° per step (before gear reduction)
* Steps per Revolution (after gearing): 4096 steps (approx. 0.088° per step)
* Torque: ~3-4 mN\*m

**Why It's Ideal for This Project:**
1.  **Precision Movement:** The high number of steps per revolution (4096 steps) allows for smooth and accurate control of the model's rotational axes, essential for demonstrating the precise movements of a real C-arm system.
2.  **Low Power Requirement:** Operating at 5V with a relatively low current draw makes it compatible with the Arduino Nano ESP32 and suitable for battery-powered or USB-powered operation.
3.  **Built-in Gear Reduction:** The 64:1 internal gearbox provides increased torque, ensuring that even with the lightweight 3D-printed components, the motor can move the structure with stability.
4.  **Quiet and Controlled Motion:** Unlike DC motors, stepper motors provide controlled, incremental movement without requiring feedback systems like encoders, making them simpler to implement in this application.

#### Motor Driver: ULN2003
To drive the 28BYJ-48 stepper motor, the ULN2003 motor driver board was selected. This board includes the ULN2003 Darlington transistor array, designed to efficiently handle the inductive loads of stepper motors.

**Technical Specifications:**
* Voltage Input: 5V-12V
* Max Output Current per Channel: 500mA
* Darlington Transistor Configuration: Provides amplified current to drive the stepper motor coils
* 4-Phase LED Indicators: Useful for debugging and monitoring motor operation

**Why It's Ideal for This Project:**
1.  **Simplifies Stepper Control:** The ULN2003 board simplifies motor control by allowing direct connection to an Arduino's digital I/O pins, reducing the need for complex circuitry.
2.  **Current Amplification:** Since the Arduino cannot directly supply enough current to drive the stepper motor, the ULN2003 ensures the motor receives sufficient power.
3.  **Compact and Efficient:** The small size of the driver board makes it easy to integrate into the model without adding unnecessary bulk.
4.  **Integrated LED Indicators:** Helps during development and troubleshooting by visually displaying motor coil activations.

#### Microcontroller: Arduino Nano ESP32
The Arduino Nano ESP32 was chosen as the main controller for the system due to its compact size, wireless communication capabilities (Wi-Fi & Bluetooth), and sufficient processing power for motor control and remote operation. We were also familiar with using ESP32 controllers from our MSE-2202 course.

**Technical Specifications:**
* Processor: ESP32-based (Dual-core Tensilica LX6)
* Clock Speed: 240 MHz
* Flash Memory: 16MB
* RAM: 512KB
* I/O Pins: 22 GPIO pins
* Wireless Connectivity: Wi-Fi 802.11 b/g/n and Bluetooth 5.0
* Operating Voltage: 3.3V (5V tolerant I/O)

**Why It's Ideal for This Project:**
1.  **Wireless Control Capability:** The Wi-Fi and Bluetooth features enable seamless communication with a phone or laptop, allowing users to control the C-arm model remotely.
2.  **Compact and Power Efficient:** The small footprint makes it easy to integrate into the 3D-printed assembly, and its low power consumption ensures it can run on USB or battery power.
3.  **Sufficient I/O for Stepper Control:** The Arduino Nano ESP32 has enough GPIO pins to control multiple stepper motors simultaneously while also handling communication protocols for the remote interface.
4.  **Strong Community Support:** The ESP32-based Nano has extensive documentation and community support, making development and troubleshooting easier.

## Results

The mechanical components of our model were designed using SolidWorks and manufactured with a Creality K1C 3D printer. SolidWorks provided essential tools for our design process, including automatic gear generation, component importing (such as motors), and virtual assembly visualization. These features allowed us to ensure proper component fit and functionality before manufacturing.

Prioritizing ease of assembly, maintenance, and accessibility, we designed the driving components of the system to be open and visible rather than concealed, even though a real medical C-arm system typically hides its internal mechanisms. This approach enhances understanding and serviceability, making the model more suitable as a training and demonstration tool.

Our design successfully achieved motorized control of all three degrees of freedom:

* The table movement is driven by a stepper motor coupled to a rack and pinion mechanism.
* All gears in the system were designed with a 20° pressure angle and a 2mm module, ensuring proper meshing while allowing flexibility in gear sizing by adjusting the number of teeth.
* Both the table and C-arm are mounted on dovetail sliders, providing a simple yet effective way to constrain motion along the desired directions.
* The C-arm itself is externally geared and driven by a pinion attached to a stepper motor on the supporting arm, allowing smooth rotation.
* The entire C-arm assembly is mounted on a stepper motor at the main stand, enabling controlled rotation of the structure.
  
This design ensures that each axis of motion is precisely controlled, while the 3D-printed components keep the model lightweight, manufacturable, and easily reproducible.


The software (appendix) was designed to control the 3-axis allowing precise control of the vertical motion the C-arm, linear travel of the table he rotational motion of the stand using stepper motors. The goals included implementing preset locations for the motors, safe motion limits, and EEPROM based position memory, and user friendly based serial commands for live control and training. The implementation successfully supported all intended commands, including real-time directional control (c, cc), position setting via goto, home, and persistent memory using EEPROM. Each motor was assigned appropriate motion constraints to prevent mechanical over-travel, with software-enforced boundary checks via polling techniques. Emergency stop functionality allowed immediate interruption of motion, again through polling. The system preserved position data after each movement using EEPROM and restored it at startup, ensuring position accuracy across sessions unless manually moved while powered off. Presets were added for each motor, allowing repeatable positioning for common procedures. The demo mode highlighted full system capabilities in sequence, validating the system’s integration and EEPROM memory after power cycles. Commands followed a consistent motor#,command format, which made testing, debugging, and demonstrations easy to manage without additional hardware like buttons or displays. All state information, including positions and movement confirmations, were echoed clearly through serial output. The system used a lightweight serial interface to keep the design simple and platform independent, easily able to make the code open source. Overall the software met all of the goals set and helped prove the prototype designed has the potential to a feasible training tool.

The electrical hardware for the system connected successfully without any major issues. Using a breadboard allowed 5V power distribution to all three motor driver boards and created a shared ground reference for the ESP32, motor drivers, and other components, ensuring stable operation. All three stepper motors were able to move smoothly within their defined physical limits without causing any damage to the structure or components.

The ESP32 Nano board interfaced reliably with the motor drivers, and the system could consistently execute commands such as changing motor speed, moving to target positions, and stopping safely. Each motor responded accurately to inputs and demonstrated the ability to simulate realistic C-arm motion Though functionally complete, one area for improvement is wire management. The breadboard setup, while useful for prototyping, introduced clutter potential for loose connections, which wasted time when they weren’t identified immediately. For future iterations, switching to a personalized PCB would improve both reliability and professionalism.

## Discussion

During the initial design process, several technical and collaborative challenges emerged that significantly influenced both the physical construction and the overall functionality of the C-arm simulator. A primary issue arose from the limitations of 3D printing of small, detailed components. Specifically, the fabrication of compact gear teeth proved problematic, as small teeth often failed to print with enough precision to create the tooth profile. As a result, the design was modified to include larger gear teeth, which could be printed, but conflicted with the original intent of a compact form factor. This trade-off between mechanical performance and spatial efficiency necessitated iterative redesigns. Additionally, collaborative constraints within the SolidWorks posed difficulties. The platform's limited collaboration capabilities made it challenging to integrate separate components developed by different team members. These issues were addressed through improved communication and consistent coordination, including regular design reviews to ensure alignment across subsystems.

Another major design consideration involved the physical layout and appearance of the prototype. While there was an intention to show the clean, enclosed design of a real medical C-arm, certain compromises were necessary to support prototyping functionality. Electrical components such as motors, sensors, and wiring were partially exposed or placed in accessible compartments to facilitate ease of testing and maintenance—particularly since motor placement, while ideally hidden for aesthetic reasons, became problematic when access was restricted. Gears, although typically covered in the actual device, were intentionally left exposed to allow clear visualization of motion and mechanical interaction during demonstration. Lastly, a critical objective was to ensure that the simulator could be easily replicated by others. To support this, the design was modularized and structured with standardized, printable parts, enabling users to fabricate, assemble, and demonstrate the system without special tools or complex setup. This emphasis on replicability enhances the simulator’s value as an educational and outreach tool, particularly in resource-limited or training-focused environments. 

Another significant design challenge involved achieving proper mechanical fits for the sliders used on both the C-arm and the base table. These sliders are essential for enabling smooth, controlled motion that closely mimics the mobility of a real surgical C-arm. After exploring multiple configurations, a dovetail-shaped slider was ultimately selected for its inherent mechanical stability and self-aligning properties. However, fine-tuning the geometry of this design required extensive iteration. Variations in printer resolution and minor dimensional inaccuracies meant that even small changes to the dovetail profile had a significant impact on how well the parts fit and moved together. Multiple prototypes were printed and tested, with adjustments made to both the male and female components to get a balance between easy sliding and structural rigidity. The process highlighted the importance of tolerancing and material behavior in additive manufacturing and reinforced the need for precise calibration when designing mechanical interfaces for 3D-printed systems. 

During the fabrication phase, several printing-related challenges arose. One of the primary obstacles was the overhangs in the design, particularly in components with geometries with internal channels. These overhangs need the generation of supporting structures during slicing, which increased overall print time but also led to material waste. These supports had to be carefully snapped off and sanded down, which added time and came with minor surface inconsistencies that had to be manually corrected. Furthermore, due to the use of a printer with a relatively wide-diameter nozzle, there were limitations in obtaining parts of high precision. This constraint lead us to modify the geometry to accommodate slightly looser tolerances and avoid interference fits that would cause jamming.

To add to these difficulties was the limited build volume of the 3D printer. As several components of the C-arm were larger than the maximum allowable dimensions certain parts had to be split into two or more sections. These segments were printed separately and then manually joined using superglue. This process required careful consideration of where to divide the parts to preserve the load paths and component interactions. Through iterative testing and part resizing the team adapted the design to the constraints of this printing process. These adjustments not only made fabrication feasible with the available tools but also informed valuable lessons about designing for manufacturability and the practical limitations of desktop additive manufacturing in prototyping functional mechanical systems.

The last problem faced with the 3D printer was the filament clogging the nozzle. This typically occurs when small particles build up and obstruct the opening, therefore the filament runs unevenly, printing inconsistent parts. Multiple hour print jobs were halted by these issues forcing the team to spend much more time than needed waiting on parts to be completed. Through disassembling the nozzle and cleaning out the outer surface, all necessary parts were printed. Due to time limitations of the unforeseen printer clogging, a proper, professional stand was missing. To delegate this, a shoe box was used to cover all electrical wiring and provide necessary support to our parts. Although giving an unaesthetic finish, the shoe box provided necessary functionality for the in-class presentation. 

After ordering all required parts for our system, configuring the Arduino Nano Esp32 with the stepper motors and correctly connecting all the wires introduced many challenges. 

For one, the pinout of the Arduino came with ESP32 pins and Arduino NANO pins. Understanding that the NANO pins were the ones required to configure to activate the motor was a learning process. Since our prior experience had primarily involved working with ESP32 configurations, adapting to the NANO logic introduced some initial confusion. This was overcome through trial and error. Initially following the pinout for GPIO pins and inputting those values into the code meant the board wasn’t pulling the correct pins high when the code uploaded. Once changing the wiring on the board to follow the NANO pinout as opposed to the GPIO pin out, the motor efficiently powered on. Understanding which pins were firing in which order to see how they needed to be configured was the next learning curve. To energize the 4 coils of the stepper motor in a sequence that causes rotational motion rather than just vibration took some trial and error. For half stepping mode it was found that having the Arduino send signals to the IN1-IN4 pins in the order of 1-3-2-4 then wiring the digital pins from the Arduino to IN1-IN4 in the same order. By keeping the pinout consistent with our code, the direction and speed of the stepper motor is precisely controlled. 

While setting up the software infrastructure, the wireless connection between the slicer platform and the accelerometer sensors raised some complications. Ensuring a reliable Bluetooth connection proved to be a finicky process. Common issues included the connection appearing active without any corresponding sensor data or, at times, a complete failure to connect. These problems were typically resolved by restarting the application, which helped re-establish a proper link between devices. 

Throughout the prototyping process, many codes were uploaded to the Arduino to test its functionality. The first code that we uploaded to board was a standard motor driving code obtained from [FORUM, Arduino, Stepper Motor with ULN2003 Driver]3. This code sketched the motor driver setup for this specific stepper motor. This website also offered us solutions to problems we may encounter. One of which being the motor vibrating but not spinning properly. This was an indicator that we needed to find the right order of the coils to successfully sequence the steps. Looking at the codes proposed in the forum we iterated our own variation through trial and error. We successfully changed the steps to make to make a full rotation of the motor to be 2038 steps. This gave us a way to specifically preset values based on the step count for precise control of the c arm. With the wireless control configured, we were able to configure the motor positions based on user input. Such X-ray views like anteroposterior view where the c arm is vertical and the x ray passes from front to back of the patient, lateral view with a horizontal c arm passing from one side of the patient to another or oblique view with an angled arm for view of overlapping anatomy can therefore be precisely controlled and preset. 

During development process we explored connecting the ESP32 to Arduino loT cloud to enable wireless control. This would allow us to control the system form any screen such as phones, ipads, or other touch screens. The goal was to reduce wiring to simplify the set up for control via Wifi. To connect the Arduino loT cloud, the device needs an ID and Secret Key which are given when you link the board to the cloud through your computer. It also needs to access to a Wifi network needing an SSID and password. However, the university’s Wifi uses specific authentication requiring a username and password which is not supported by the default Arduino loT connection. While being able to successfully implement this outside of campus, it frequently failed to keep a stable link with the cloud service and would often bug out affecting the USB connectivity. Due to the limitations mentioned, we relied on a direct USB connection for programming to guarantee correct functionality. Throughout this process we learnt that enterprise networks like eduroam are hard to work around, and although Arduino loT Cloud is beginner friendly it lacks flexibility in handling more complex authentication. We understood to design for the environment that the device is deployed in, not only the device itself.

Throughout this project lots of experience was gained to learn how to implement and design an embedded multi-motor control system. Familiarity with the ESP32 from previous course work allowed for more advanced control of the system through the software development and sets this prototype apart from others which have similar mechanical designs. This included learning how to add firmware functionality like using, and accessing the EEPROM to preserve state of the system, motion limits and serial feedback also taught the importance of safety and user interface clarity in system design, and translating rotational motion into linear distance. It was a learning curve to step into the mind of users and what they would need or want in a device like this, thinking about these objectives lead to the decisions regarding what functions the software would have like repeatability and safety. 

Preset positions are an important feature because they allow movement of all the axis to a consistent and repeatable location that can be found multiple times. In real life this would be really useful for training new technicians or if they were looking at the same patient multiple times and needed to use a similar setup each time. 

The goto command moves the C-arm to a specific linear position in mm simulating precise motorized targeting in fluoroscopy equipment. Specific positions can be recalled enabling repeatable and accurate positioning of the imaging system, similar to functionality of a clinical C-arm. Pre-set positioning also allows fast and efficient repeatability of each of the motor axis allowing Analogous the zeroing of medical equipment, each motor can return to a preset home position. This feature ensures that the relative position of the motors is known and has a feature to be zeroed to define and save a new home position if calibration or realignment of system is needed after servicing or repositioning.

The pos command displays the position of all the motors or of an individual motor to tell the user the systems relative location to its home position. The system can output both step count and converted linear position (in mm) for each motor, which is helpful during setup, calibration, or debugging. Limits on the linear and rotational motion of the motors automatically halts motion beyond defined bounds, protecting the physical structure from damage or damage to connected components and emulating safety limits used in clinical systems.

One major area of learning development included learning how to program EEPROM to store positional data. In the event of power loss, medical equipment must retain its last known state to avoid misalignment or unsafe motion upon reboot. To address this, the firmware stores the current position of each motor in EEPROM whenever a move is completed. The position in either linear distance, so mm, or in degrees for the rotation of the stand, as well as the step count, all relative to the calibrated home position is saved. Upon startup, the system loads these positions, allowing the simulator to resume operation without requiring manual reset or recalibration. This is especially important in medical contexts, where exceeding limits due to lost calibration data could result in equipment damage or patient harm. 

As Safety is a critical consideration in both real and simulated medical systems stop command has been implemented to immediately halt all motor movement. This is monitored continuously using a polling technique. In a full clinical device, this behavior would typically be managed through a hardware interrupt, in the form of an Emergency-Stop button, triggering an interrupt service routine (ISR) to override normal operation and bring the system to a safe stop.

One challenge was the lack of motor feedback, the system relied solely on software tracking. As a result, position values could become inaccurate if the motors were moved while powered off. This was addressed with a zero command, but the user needs to be aware that this calibration must be completed. Another challenge was ensuring EEPROM saved data consistently on the Nano ESP32 platform, which required careful use of EEPROM.commit(), something that is specific to the ESP32 to successfully save the data. Figuring out how to track the position was one of the biggest struggles because any backlash, vibration from motors or steps that were missed or skipped resulted in inaccuracy and errors that accumulated and made locating much more difficult. To improve the system further, I would consider adding rotary encoders to each motor axis. This would allow closed-loop feedback and eliminate positional drift or inaccuracies from potential missed steps or manual displacement. The use of external hardware interrupts, like for an emergency-stop button, would also enhance real-world realism and safety compliance. Learning how to integrate these features would solve a lot of limitations the design has currently. Overall, this project provided a lot of insight about how to think like a system designer, balance mechanical design, hardware, software and user experience.


## Conclusion

We have prototyped a solution conveying a C-arm Fluoroscopy machine which generates real time medical imaging for ease of training and demonstration purposes. This accessible surgical simulator creates more accessible opportunities for learning while keeping a cost-efficient structure. This model prioritizes safety of use, which reduces the need for radiation exposure during the learning process. In a societal point of view, this improves patient safety by allowing more efficient training and makes advances in surgical assistance more inclusive for regions which are more under-resourced. It encourages medical technology advancements and innovations. 

Throughout this term we collectively developed a deeper understanding of the mechanics of the device, as well as sensor integration and translating a system into a compact but accurate model. This has shown us that effective surgical training does not always require the full-scale tangible appliance. 

Through SolidWorks, we learned to integrate all components together ensuring proper movement of all parts, as well as incorporating space for electrical components. Finding the right balance between ease of assembly and minimizing the space taken up by the model was a learning process. Considering the rotation of the motors, the placement of sensors, and the integration of the microcontroller presented a unique set of challenges. Each miscalculation became an opportunity to troubleshoot and refine our approach, ultimately providing valuable insights into system design and problem-solving. 

Balancing realism and an abstract concept of a device to accurately simulate motion of a C-arm without the need of the physical machine was an overall positive task. This came with extensive teamwork and communication. Explaining technical processes to peers which different practical backgrounds, collaboration between group members and troubleshooting when problems arose were pivotal notions into our success.
